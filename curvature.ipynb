# TORCS Training and Validation Tracks - Visualizing the Curvature

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from gym_torcs.curvature import RacingLineCurvatureMap
from gym_torcs.constants import RACING_LINES, CURVATURE
```

```python
TRACKS = {
    "alpine-1": "Alpine-1 (Train)",
    "alpine-2": "Alpine-2 (Val)",
    "g-track-2": "G-Track-2 (Val)",
    "street-1": "Street-1 (Train)",
    "aalborg": "Aalborg (Train)",
    "forza": "Forza (Train)",
    "brondehach": "Brondehach (Val)",
    "wheel-1": "wheel-1 (Train)",
    "wheel-2": "wheel-2 (Train)",

}

TRACK_ORDER = [
    # Training tracks 
    "alpine-1",
    "aalborg",
    "street-1",
    "forza",
    "wheel-1",
    "wheel-2",

    # Validation tracks
    "g-track-2",
    "brondehach",
    "alpine-2",
]
```

```python
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(12, 8))
axes = axes.flatten()

for track_name, ax in zip(TRACK_ORDER, axes):
    racing_line = RacingLineCurvatureMap.from_csv(RACING_LINES[track_name]) 

    x, y = racing_line.x, racing_line.y
    curvature = racing_line.curvature / CURVATURE

    scatter = ax.scatter(x=x, y=y, c=curvature, cmap="viridis")
    ax.set_title(TRACKS[track_name]) 
    ax.axis("off")

    cbar = fig.colorbar(scatter, ax=ax)
    cbar.set_label("Curvature")
```
